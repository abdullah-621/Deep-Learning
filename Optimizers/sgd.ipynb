```python
import keras
from keras import layers, optimizers

# 1. Define your model
model = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(10,)),
    layers.Dense(1, activation='sigmoid')
])

optimizer = optimizers.SGD(learning_rate=0.01, momentum=0.0, nesterov=False)  # only SGD
optimizer = optimizers.SGD(learning_rate=0.01, momentum=0.9, nesterov=False)  # SGD + momentum
optimizer = optimizers.SGD(learning_rate=0.01, momentum=0.9, nesterov=True)  # SGD + momentum + nesterov

# 3. Compile the model
model.compile(optimizer=optimizer, loss='mse')
```
