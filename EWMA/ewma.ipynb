```python
df['ewma'] = 0.0
```

```python
df.head()
```

```python
def ewma(df):
  beta = 0.9
  
  ewma_values = [df['meantemp'].iloc[0]]

  for i in range(1, df.shape[0]):
    Vt = (beta * ewma_values[i - 1]) + (1 - beta) * df['meantemp'].iloc[i]
    ewma_values.append(Vt)

  df['ewma'] = ewma_values
  
  return df
```

```python
df = ewma(df)
```
